```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
rows_changed = {}

CATEGORY_MAP = {
    'Emergency Supplies': 'Emergency',
    'Groceries': 'Grocery',
    'Games & Entertainment': 'Entertainment',
    'Household Essentials': 'Household',
    'Baby Products': 'Baby',
    'Breakfast Foods': 'Breakfast',
    'Alcoholic Beverages': 'Alcohol',
    'Health & Wellness': 'Health',
}


def clean_store_id(col):
    col = col.str.upper().str.strip()
    col = col.str.replace('-', '').str.replace('_', '').str.replace(' ', '')
    col = 'FL-' + col.str[2:]
    return col


def clean_transactions(df):
    rows_changed['duplicates'] = df.duplicated().sum()
    df = df.drop_duplicates().copy()

    df['unit_price'] = df['unit_price'].str.replace('$', '')
    df['unit_price'] = df['unit_price'].astype(float)
    rows_changed['unit_price'] = len(df)

    rows_changed['quantity'] = (df['quantity'] < 0).sum()
    df = df[df['quantity'] >= 1].copy()

    df['timestamp'] = pd.to_datetime(df['timestamp'], format='mixed', errors='coerce')
    rows_changed['timestamp'] = df['timestamp'].isna().sum()
    df = df[df['timestamp'].notna()].copy()

    before = df['store_id'].copy()
    df['store_id'] = clean_store_id(before)
    rows_changed['store_id'] = int(before.ne(df['store_id']).sum())

    before = df['category'].copy()
    df['category'] = (before.str.strip().str.title().replace(CATEGORY_MAP))
    rows_changed['category'] = int(before.ne(df['category']).sum())

    return df


def clean_stores(df):
    df = df.copy()
    df['store_id'] = clean_store_id(df['store_id'])

    df['state'] = (
        df['state'].str.strip().str.upper()
        .replace({'FLORIDA': 'FL', 'FLA.': 'FL'})
    )
    df['region'] = df['region'].str.strip().str.title()

    rows_changed['stores'] = int(
        df.duplicated('store_id').sum()
    )
    df = df.drop_duplicates('store_id').copy()

    df['has_coordinates'] = (
        df[['latitude', 'longitude']].notna().all(axis=1)
    )
    return df


transactions_raw = pd.read_csv('../data/walmart_transactions.csv')
stores_raw = pd.read_csv('../data/store_locations.csv')

transactions_typed = clean_transactions(transactions_raw)
stores = clean_stores(stores_raw)

print('rows before:', len(transactions_raw))
print('rows after:', len(transactions_typed))
print('rows removed:', len(transactions_raw) - len(transactions_typed))
print('store_ids:', transactions_typed['store_id'].nunique())
print('categories:', transactions_typed['category'].nunique())
print('stores in sales but not in the store file:', set(transactions_typed['store_id']) - set(stores['store_id']))
```

```python
import requests, pandas as pd, time

BASE = 'https://archive-api.open-meteo.com/v1/archive'
RETRY_STATUSES = {429, 500, 502, 503, 504}

def get_weather(start, end, lat=28.54, lon=-81.38, tries=3):
    if tries < 1:
        raise ValueError('tries must be at least 1')
    params = {
        'latitude': lat, 'longitude': lon,
        'start_date': start, 'end_date': end,
        'daily': 'precipitation_sum,windspeed_10m_max,pressure_msl_mean',
        'timezone': 'America/New_York',
    }
    for attempt in range(1, tries + 1):
        try:
            r = requests.get(BASE, params=params, timeout=10)
        except (requests.Timeout, requests.ConnectionError):
            if attempt == tries:
                raise
            time.sleep(2 ** attempt)
            continue

        if r.status_code in RETRY_STATUSES:
            if attempt == tries:
                r.raise_for_status()
            time.sleep(2 ** attempt)
            continue

        r.raise_for_status()
        return pd.DataFrame(r.json()['daily'])


wx = get_weather('2024-09-01', '2024-09-30')

wx = wx.rename(columns={
    'time': 'date',
    'precipitation_sum': 'precip_mm',
    'windspeed_10m_max': 'wind_max_kmh',
    'pressure_msl_mean': 'pressure_hpa',
})
wx['date'] = pd.to_datetime(wx['date'])
wx['source'] = 'open-meteo'

print('rows:', len(wx))
wx.head()
```

### Q1 — Demand surge

**TODO:** your baseline window, your surge window, the average for each, and the multiple. Plus the chart. State the window boundaries explicitly — a multiple without a window is not a finding.

```python
sales['date'] = sales['timestamp'].dt.normalize()
q1_daily = sales.groupby(['date', 'category'])['quantity'].sum().unstack(fill_value=0)

baseline = q1_daily.loc['2024-09-08':'2024-09-15'].mean()
surge = q1_daily.loc['2024-09-19':'2024-09-25'].mean()

q1_results = pd.DataFrame({
    'baseline units per day': baseline,
    'surge units per day': surge
})

q1_results['multiple'] = surge / baseline
q1_results['percent change'] = (q1_results['multiple'] - 1) * 100

q1_results = q1_results.sort_values('multiple', ascending=False)

display(q1_results.round(2))

top5 = q1_results.head(5).index

ax = q1_daily[top5].plot(figsize=(12, 5))

ax.axvspan(pd.Timestamp('2024-09-19'), pd.Timestamp('2024-09-25'), color='orange', alpha=0.12, label='Surge window')
ax.axvline(pd.Timestamp('2024-09-26'), color='gray', linestyle='--', label='Landfall (Sept 26)')

ax.set_title('Breakfast sales rose 5.2x in the week before landfall')
ax.set_xlabel('Date')
ax.set_ylabel('Units sold')
plt.legend()
plt.tight_layout()
plt.show()
```

**Q1 answer, in one sentence a manager could read:** Sales increased during the pre-landfall week.

```python
units_before = sales.groupby('sku')['quantity'].sum().sum()
units_after = sales.groupby('canonical_sku')['quantity'].sum().sum()
assert units_before == units_after

pt = sales[sales['canonical_sku'] == 'POPTART-STRAWBERRY'].copy()
pt['date'] = pt['timestamp'].dt.normalize()

before = pt.pivot_table(index='date', columns='sku', values='quantity', aggfunc='sum', fill_value=0)
after = before.sum(axis=1)

fig, axes = plt.subplots(1, 2, figsize=(12, 5), sharey=True)

before.plot(ax=axes[0], title='Before: four SKUs, none above 240 units a day')
after.plot(ax=axes[1], title='After: one product peaking at 678 units on Sept 25')

for ax in axes:
    ax.set_xlabel('Date')
    ax.set_ylabel('Units sold')

plt.tight_layout()
plt.show()

print('Original SKUs:', sales['sku'].nunique())
print('Canonical products:', sales['canonical_sku'].nunique())
print('Pop-Tarts variants:', pt['sku'].nunique())
print('Units before:', units_before)
print('Units after:', units_after)
```

### Q3 — Weather correlation

**TODO:** the API call, the validated join, and the relationship you found. Include:

- the row count before and after the join, proving nothing was lost
- whether the data is live or fallback
- how many days the correlation is computed over

If your n is small, say so here rather than being asked about it on Monday.

```python
orlando = sales[sales['store_id'] == 'FL-239'].copy()
orlando['date'] = orlando['timestamp'].dt.normalize()

daily_orlando = orlando.groupby('date', as_index=False)['quantity'].sum().rename(columns={'quantity': 'units_sold'})

weather_sales = daily_orlando.merge(wx, on='date', how='left', validate='one_to_one', indicator=True)

assert len(weather_sales) == len(daily_orlando)

valid = weather_sales.dropna(subset=['units_sold', 'wind_max_kmh']).sort_values('date')

print('Rows before join:', len(daily_orlando))
print('Rows after join:', len(weather_sales))
print('Matched days:', len(valid))
print('Unmatched days:', (weather_sales['_merge'] != 'both').sum())
print('Weather source:', valid['source'].unique())
print('Wind-sales correlation:', round(valid['units_sold'].corr(valid['wind_max_kmh']), 3))

for lag in range(5):
    r = valid['units_sold'].corr(valid['wind_max_kmh'].shift(-lag))
    print(f'Sales vs wind {lag} days later: {r:.3f}')

plt.figure(figsize=(7, 4))
plt.scatter(valid['wind_max_kmh'], valid['units_sold'])
plt.xlabel('Maximum daily wind speed (km/h)')
plt.ylabel('Units sold')
plt.title('Orlando sales rise on windier days (r = 0.47, n = 30)')
plt.tight_layout()
plt.show()
```

**Q3 answer, with its caveat:** Our data contains daily wind speed data across the month compared to sales with 30 matching rows before and after the join. The data is live indicated by the open-meteo line and in computed over 30 days.
